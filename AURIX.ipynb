```python
import requests
from tqdm.auto import tqdm

url = "https://zenodo.org/records/19336329/files/dev_bearingEmu.zip"
zip_path = DATA_DIR / "dev_bearingEmu.zip"

with requests.get(url, stream=True, timeout=60) as response:
    response.raise_for_status()
    total = int(response.headers.get("content-length", 0))

    with open(zip_path, "wb") as f, tqdm(
        total=total,
        unit="B",
        unit_scale=True,
        desc="Downloading AURIX dataset"
    ) as progress:
        for chunk in response.iter_content(chunk_size=1024 * 1024):
            if chunk:
                f.write(chunk)
                progress.update(len(chunk))

print("Download complete:", zip_path)
print("File size (GB):", round(zip_path.stat().st_size / (1024**3), 2))
```

```python

from pathlib import Path
import numpy as np
import librosa
import soundfile as sf

wav = Path(
    "/content/AURIX_data/bearingEmu/train/"
    "section_00_source_train_normal_0067_noAttribute.wav"
)

audio, sr = sf.read(wav, dtype="float32")
assert audio.ndim == 2 and audio.shape[1] == 2
assert sr == 16000

def logmel(signal):
    mel = librosa.feature.melspectrogram(
        y=signal,
        sr=sr,
        n_fft=1024,
        hop_length=512,
        n_mels=128,
        power=2.0,
    )
    return 10.0 * np.log10(np.maximum(mel, 1e-10)).T

def make_vectors(features, frames=5):
    # features: time × frequency
    if len(features) < frames:
        raise ValueError("Not enough frames")
    windows = np.stack(
        [features[i:i + frames].reshape(-1)
         for i in range(len(features) - frames + 1)]
    )
    return windows

m0 = logmel(audio[:, 0])
m1 = logmel(audio[:, 1])

# Same preprocessing per channel; early fusion joins features per frame.
v0 = make_vectors(m0)
v1 = make_vectors(m1)
vf = make_vectors(np.concatenate([m0, m1], axis=1))

for name, vectors in [
    ("Channel 0", v0),
    ("Channel 1", v1),
    ("Early fusion", vf),
]:
    print(
        f"{name}: shape={vectors.shape}, "
        f"finite={np.isfinite(vectors).all()}"
    )

assert v0.shape[1] == 640
assert v1.shape[1] == 640
assert vf.shape[1] == 1280
assert all(np.isfinite(v).all() for v in (v0, v1, vf))

print("\nFrame-stacking check: PASS")
```

```python
from pathlib import Path
import numpy as np
import librosa
import soundfile as sf

wav_path = Path(
    "/content/AURIX_data/bearingEmu/train/"
    "section_00_source_train_normal_0000_noAttribute.wav"
)

audio, sr = sf.read(str(wav_path), dtype="float32")
assert audio.ndim == 2 and audio.shape[1] == 2
assert sr == 16000

def log_mel(channel_audio, sr, n_mels=128, n_fft=1024, hop_length=512):
    mel = librosa.feature.melspectrogram(
        y=channel_audio,
        sr=sr,
        n_mels=n_mels,
        n_fft=n_fft,
        hop_length=hop_length,
        power=2.0,
        fmin=0.0,
        fmax=None,
    )
    return 10.0 * np.log10(np.maximum(mel, np.finfo(float).eps))

mel0 = log_mel(audio[:, 0], sr)
mel1 = log_mel(audio[:, 1], sr)
fused = np.concatenate([mel0, mel1], axis=0)

for name, features in [
    ("Channel 0", mel0),
    ("Channel 1", mel1),
    ("Early fusion", fused),
]:
    print(
        f"{name}: shape={features.shape}, "
        f"finite={np.isfinite(features).all()}, "
        f"mean={features.mean():.4f}, std={features.std():.4f}"
    )

assert mel0.shape == mel1.shape
assert fused.shape[0] == 2 * mel0.shape[0]
assert all(np.isfinite(x).all() for x in [mel0, mel1, fused])

print("\nThree-view feature extraction: PASS")
```

```python

from pathlib import Path
import csv
import random

data_dir = Path("/content/AURIX_data/bearingEmu/train")
out_dir = Path("/content/AURIX.V1/experiments")
manifest = out_dir / "recording_split_seed13711.csv"
seed = 13711

if not data_dir.exists():
    raise FileNotFoundError(f"Training directory not found: {data_dir}")

files = sorted(data_dir.glob("*.wav"))
if len(files) != 1000:
    raise ValueError(f"Expected 1000 training WAVs, found {len(files)}")

# Confirm that this experiment uses normal training recordings only.
if any("_anomaly_" in f.name.lower() for f in files):
    raise ValueError("Anomaly recording found in training directory.")

groups = {"source": [], "target": []}

for f in files:
    name = f.name.lower()
    if "_source_train_normal_" in name:
        groups["source"].append(f.name)
    elif "_target_train_normal_" in name:
        groups["target"].append(f.name)
    else:
        raise ValueError(f"Unrecognized training filename: {f.name}")

if len(groups["source"]) != 990 or len(groups["target"]) != 10:
    raise ValueError(
        f"Unexpected domain counts: "
        f"source={len(groups['source'])}, target={len(groups['target'])}"
    )

# Keep the split deterministic and separate by domain.
rng = random.Random(seed)
rows = []

for domain, names in groups.items():
    names = names.copy()
    rng.shuffle(names)

    # 10% source validation; 2 of the 10 target recordings for validation.
    n_valid = round(0.10 * len(names))
    if domain == "target":
        n_valid = 2

    valid_names = set(names[:n_valid])

    for name in sorted(names):
        rows.append({
            "filename": name,
            "domain": domain,
            "split": "validation" if name in valid_names else "train",
            "label": "normal",
            "seed": seed,
        })

# Verify no recording can occur in both splits.
train_names = {r["filename"] for r in rows if r["split"] == "train"}
valid_names = {r["filename"] for r in rows if r["split"] == "validation"}

assert train_names.isdisjoint(valid_names)
assert len(train_names) + len(valid_names) == 1000
assert len(valid_names) == 101
assert sum(r["split"] == "validation" and r["domain"] == "target"
           for r in rows) == 2

out_dir.mkdir(parents=True, exist_ok=True)

if manifest.exists():
    # Preserve an existing experiment artifact rather than silently replacing it.
    with manifest.open(newline="", encoding="utf-8") as f:
        existing = list(csv.DictReader(f))
    if len(existing) != 1000:
        raise ValueError(f"Existing manifest has {len(existing)} rows; inspect it before replacing.")
    print(f"Existing manifest preserved: {manifest}")
else:
    with manifest.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(
            f, fieldnames=["filename", "domain", "split", "label", "seed"]
        )
        writer.writeheader()
        writer.writerows(rows)
    print(f"Created manifest: {manifest}")

print("\nExpected split:")
print(f"Train:      {len(train_names)} recordings")
print(f"Validation: {len(valid_names)} recordings")
print(f"Source validation: 99")
print(f"Target validation: 2")
print("Test files accessed: 0")
print("Recording-level overlap: 0")
```

```python

from pathlib import Path
import csv
import numpy as np
import soundfile as sf
import librosa

root = Path("/content/AURIX_data/bearingEmu/train")
manifest = Path("/content/AURIX.V1/experiments/recording_split_seed13711.csv")

with manifest.open(newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

train_rows = [r for r in rows if r["split"] == "train"]
valid_rows = [r for r in rows if r["split"] == "validation"]

assert len(rows) == 1000
assert not ({r["filename"] for r in train_rows}
            & {r["filename"] for r in valid_rows})

def extract_views(path):
    audio, sr = sf.read(str(path), dtype="float32")
    assert sr == 16000 and audio.ndim == 2 and audio.shape[1] == 2

    views = []
    for channel in (0, 1):
        mel = librosa.feature.melspectrogram(
            y=audio[:, channel],
            sr=sr,
            n_mels=128,
            n_fft=1024,
            hop_length=512,
            power=2.0,
        )
        log_mel = 10 * np.log10(np.maximum(mel, np.finfo(np.float32).eps))
        views.append(log_mel)

    assert views[0].shape == views[1].shape
    fused = np.concatenate(views, axis=0)

    return views[0], views[1], fused

samples = [
    ("train", train_rows[0]),
    ("train", train_rows[1]),
    ("validation", valid_rows[0]),
    ("validation", valid_rows[1]),
]

for split, row in samples:
    path = root / row["filename"]
    if not path.is_file():
        raise FileNotFoundError(path)

    ch0, ch1, fused = extract_views(path)

    print(
        f"{split:10s} | {row['domain']:6s} | {path.name}\n"
        f"  ch0={ch0.shape}, ch1={ch1.shape}, fusion={fused.shape}\n"
        f"  finite={all(np.isfinite(x).all() for x in (ch0, ch1, fused))}"
    )

print("\nManifest and multi-recording feature smoke test: PASS")
print("No model training performed.")
```

```python

from pathlib import Path
import csv, json, random
import numpy as np
import soundfile as sf
import librosa
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

SEED = 13711
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.set_num_threads(2)
device = torch.device("cpu")

data_dir = Path("/content/AURIX_data/bearingEmu/train")
exp_dir = Path("/content/AURIX.V1/experiments")
manifest = exp_dir / "recording_split_seed13711.csv"
exp_dir.mkdir(parents=True, exist_ok=True)

with manifest.open(newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

train_rows = [
    r for r in rows
    if r["split"] == "train" and r["domain"] == "source"
]
valid_rows = [
    r for r in rows
    if r["split"] == "validation" and r["domain"] == "source"
]

rng = random.Random(SEED)
rng.shuffle(train_rows)
rng.shuffle(valid_rows)
train_rows = train_rows[:20]
valid_rows = valid_rows[:5]

def channel0_vectors(filename):
    path = data_dir / filename
    audio, sr = sf.read(str(path), dtype="float32")
    if sr != 16000 or audio.ndim != 2 or audio.shape[1] != 2:
        raise ValueError(f"Unexpected audio format: {filename}")

    mel = librosa.feature.melspectrogram(
        y=audio[:, 0], sr=sr, n_mels=128,
        n_fft=1024, hop_length=512, power=2.0
    )
    log_mel = 10 * np.log10(
        np.maximum(mel, np.finfo(np.float32).eps)
    )

    # Each example contains 5 consecutive frames: 128 x 5 = 640.
    windows = np.stack(
        [log_mel[:, i:i+5].reshape(-1)
         for i in range(log_mel.shape[1] - 4)],
        axis=0
    )
    return windows.astype(np.float32)

print("Extracting pilot training features...")
train_arrays = [channel0_vectors(r["filename"]) for r in train_rows]
print("Extracting pilot validation features...")
valid_arrays = [channel0_vectors(r["filename"]) for r in valid_rows]

# Normalize using training data only to avoid validation leakage.
train_raw = np.concatenate(train_arrays)
mean = train_raw.mean(axis=0, keepdims=True)
std = train_raw.std(axis=0, keepdims=True)
std[std < 1e-6] = 1.0

train_x = (train_raw - mean) / std
valid_x = [(x - mean) / std for x in valid_arrays]

train_tensor = torch.from_numpy(train_x)
loader = DataLoader(
    TensorDataset(train_tensor),
    batch_size=256,
    shuffle=True,
    generator=torch.Generator().manual_seed(SEED)
)

model = nn.Sequential(
    nn.Linear(640, 128),
    nn.ReLU(),
    nn.Linear(128, 32),
    nn.ReLU(),
    nn.Linear(32, 128),
    nn.ReLU(),
    nn.Linear(128, 640)
).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.MSELoss()

model.train()
batch_losses = []
for (batch,) in loader:
    optimizer.zero_grad()
    reconstructed = model(batch)
    loss = criterion(reconstructed, batch)
    loss.backward()
    optimizer.step()
    batch_losses.append(loss.item())

model.eval()
record_scores = []
with torch.no_grad():
    for row, x in zip(valid_rows, valid_x):
        x_tensor = torch.from_numpy(x)
        pred = model(x_tensor)
        score = ((pred - x_tensor) ** 2).mean(dim=1).mean().item()
        record_scores.append({
            "filename": row["filename"],
            "domain": row["domain"],
            "split": "validation",
            "label": "normal",
            "mean_reconstruction_error": score
        })

tag = "pilot_channel0_seed13711"
torch.save({
    "model_state_dict": model.state_dict(),
    "normalization_mean": mean,
    "normalization_std": std,
    "input_dim": 640,
    "seed": SEED,
    "train_recordings": [r["filename"] for r in train_rows],
    "validation_recordings": [r["filename"] for r in valid_rows],
}, exp_dir / f"{tag}.pt")

with (exp_dir / f"{tag}_validation_scores.csv").open(
    "w", newline="", encoding="utf-8"
) as f:
    writer = csv.DictWriter(f, fieldnames=list(record_scores[0].keys()))
    writer.writeheader()
    writer.writerows(record_scores)

summary = {
    "experiment": tag,
    "device": str(device),
    "epochs": 1,
    "train_recordings": len(train_rows),
    "validation_recordings": len(valid_rows),
    "train_feature_vectors": int(len(train_x)),
    "mean_train_batch_mse": float(np.mean(batch_losses)),
    "mean_validation_recording_error": float(
        np.mean([r["mean_reconstruction_error"] for r in record_scores])
    ),
    "warning": "Pilot only; not a final anomaly-detection result."
}
with (exp_dir / f"{tag}_summary.json").open("w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

print("\nPILOT TRAINING COMPLETE")
print(json.dumps(summary, indent=2))
print("Saved artifacts in:", exp_dir)
```

```python

from pathlib import Path
import csv
import json

import numpy as np
import soundfile as sf
import librosa
import torch
from torch import nn
from sklearn.metrics import roc_auc_score, average_precision_score

ROOT = Path("/content/AURIX_data/bearingEmu")
OUT = Path("/content/AURIX.V1/experiments")
TEST = ROOT / "test"

MODELS = {
    "channel0": OUT / "pilot_channel0_seed13711.pt",
    "channel1": OUT / "pilot_channel1_seed13711.pt",
    "early_fusion": OUT / "pilot_earlyfusion_seed13711.pt",
}

# Reconstruct the exact model architecture used by each saved pilot.
def build_model(input_dim):
    if input_dim == 640:
        return nn.Sequential(
            nn.Linear(640, 128), nn.ReLU(),
            nn.Linear(128, 32), nn.ReLU(),
            nn.Linear(32, 128), nn.ReLU(),
            nn.Linear(128, 640)
        )
    elif input_dim == 1280:
        return nn.Sequential(
            nn.Linear(1280, 128), nn.ReLU(),
            nn.Linear(128, 32), nn.ReLU(),
            nn.Linear(32, 128), nn.ReLU(),
            nn.Linear(128, 1280)
        )
    raise ValueError(f"Unexpected input dimension: {input_dim}")

def extract_features(path, mode):
    audio, sr = sf.read(str(path), dtype="float32")
    assert sr == 16000 and audio.ndim == 2 and audio.shape[1] == 2

    channels = (0,) if mode == "channel0" else (1,) if mode == "channel1" else (0, 1)
    mels = []

    for channel in channels:
        mel = librosa.feature.melspectrogram(
            y=audio[:, channel],
            sr=sr,
            n_mels=128,
            n_fft=1024,
            hop_length=512,
            power=2.0
        )
        logmel = 10 * np.log10(
            np.maximum(mel, np.finfo(np.float32).eps)
        )
        mels.append(logmel)

    spec = mels[0] if len(mels) == 1 else np.concatenate(mels, axis=0)

    features = np.stack([
        spec[:, i:i+5].T.reshape(-1)
        for i in range(spec.shape[1] - 4)
    ]).astype(np.float32)

    expected_dim = 640 if mode in ("channel0", "channel1") else 1280
    assert features.shape[1] == expected_dim
    assert np.isfinite(features).all()
    return features

torch.set_num_threads(2)

test_files = sorted(TEST.glob("*.wav"))
assert len(test_files) == 200, f"Expected 200 test files, found {len(test_files)}"

results = []

for model_name, model_path in MODELS.items():
    assert model_path.is_file(), f"Missing model file: {model_path.name}"
    print(f"\nEvaluating {model_name}...")

    checkpoint = torch.load(
        model_path, map_location="cpu", weights_only=False
    )

    model = build_model(int(checkpoint["input_dim"]))
    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    # Handle either NumPy arrays or PyTorch tensors safely.
    mean = np.asarray(
        checkpoint["normalization_mean"].cpu().numpy()
        if torch.is_tensor(checkpoint["normalization_mean"])
        else checkpoint["normalization_mean"],
        dtype=np.float32
    )
    std = np.asarray(
        checkpoint["normalization_std"].cpu().numpy()
        if torch.is_tensor(checkpoint["normalization_std"])
        else checkpoint["normalization_std"],
        dtype=np.float32
    )
    std = np.maximum(std, 1e-6)

    rows = []

    for i, path in enumerate(test_files, start=1):
        features = extract_features(path, model_name)
        x = torch.from_numpy(((features - mean) / std).astype(np.float32))

        with torch.no_grad():
            score = float(((model(x) - x) ** 2).mean(dim=1).mean().item())

        filename = path.name.lower()
        label = 1 if "_anomaly_" in filename else 0
        domain = "target" if "_target_" in filename else "source"

        rows.append({
            "filename": path.name,
            "domain": domain,
            "label": label,
            "anomaly_score": score
        })

        if i % 50 == 0:
            print(f"  Scored {i}/{len(test_files)} recordings")

    score_file = OUT / f"pilot_{model_name}_test_scores.csv"
    with score_file.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=[
            "filename", "domain", "label", "anomaly_score"
        ])
        writer.writeheader()
        writer.writerows(rows)

    for domain in ("source", "target", "all"):
        subset = rows if domain == "all" else [
            r for r in rows if r["domain"] == domain
        ]

        y = np.array([r["label"] for r in subset], dtype=int)
        scores = np.array([r["anomaly_score"] for r in subset], dtype=float)

        if len(np.unique(y)) != 2:
            raise ValueError(f"Both classes are required for {domain} metrics.")

        results.append({
            "model": model_name,
            "domain": domain,
            "n": len(subset),
            "normal_count": int(np.sum(y == 0)),
            "anomaly_count": int(np.sum(y == 1)),
            "roc_auc": float(roc_auc_score(y, scores)),
            "pauc_max_fpr_0.1": float(
                roc_auc_score(y, scores, max_fpr=0.1)
            ),
            "average_precision": float(
                average_precision_score(y, scores)
            )
        })

report = OUT / "pilot_test_metrics_exploratory.json"
report.write_text(json.dumps(results, indent=2), encoding="utf-8")

print("\n" + "=" * 72)
print("EXPLORATORY PILOT TEST EVALUATION COMPLETE")
print("=" * 72)

for r in results:
    print(
        f"{r['model']:12s} | {r['domain']:6s} | n={r['n']:3d} | "
        f"ROC-AUC={r['roc_auc']:.4f} | "
        f"pAUC={r['pauc_max_fpr_0.1']:.4f} | "
        f"AP={r['average_precision']:.4f}"
    )

print("\nSaved metrics:", report)
print("Original model checkpoints were not modified.")
print("IMPORTANT: These are exploratory pilot results, not final independent results.")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import librosa

# Paths
project_dir = Path("/content/AURIX.V1")
data_dir = Path("/content/AURIX_data/bearingEmu/train")
experiments_dir = project_dir / "experiments"
split_path = experiments_dir / "recording_split_seed13711.csv"

# Load the existing recording-level split
df = pd.read_csv(split_path)
df = df[
    (df["domain"] == "source") &
    (df["label"] == "normal") &
    (df["seed"] == 13711)
].copy()

train_names = df.loc[df["split"] == "train", "filename"].tolist()
val_names = df.loc[df["split"] == "validation", "filename"].tolist()

# Use the same small pilot size, deterministically
train_names = sorted(train_names)[:20]
val_names = sorted(val_names)[:5]

assert len(train_names) == 20, f"Expected 20 training recordings, got {len(train_names)}"
assert len(val_names) == 5, f"Expected 5 validation recordings, got {len(val_names)}"

def extract_two_paths(filename):
    path = data_dir / filename
    if not path.exists():
        raise FileNotFoundError(path)

    audio, sr = librosa.load(str(path), sr=16000, mono=False)
    channel0 = audio[0] if audio.ndim == 2 else audio

    mel = librosa.feature.melspectrogram(
        y=channel0,
        sr=sr,
        n_fft=1024,
        hop_length=512,
        n_mels=128,
        power=2.0,
        fmin=0,
        fmax=8000
    )

    log_mel = 10 * np.log10(
        np.maximum(mel, np.finfo(np.float32).eps)
    ).astype(np.float32)

    # Spectral path: the spectrum at each time frame
    spectral = log_mel.T

    # Temporal path: frame-to-frame spectral changes
    temporal = np.diff(log_mel, axis=1).T

    return spectral, temporal

def build_features(filenames):
    spectral_parts = []
    temporal_parts = []

    for i, filename in enumerate(filenames, start=1):
        spectral, temporal = extract_two_paths(filename)
        spectral_parts.append(spectral)
        temporal_parts.append(temporal)

        if i % 5 == 0 or i == len(filenames):
            print(f"Processed {i}/{len(filenames)} recordings")

    return (
        np.concatenate(spectral_parts, axis=0),
        np.concatenate(temporal_parts, axis=0)
    )

print("Extracting training features...")
X_spec_train, X_temp_train = build_features(train_names)

print("\nExtracting validation features...")
X_spec_val, X_temp_val = build_features(val_names)

# Fit normalization on training data ONLY
spec_mean = X_spec_train.mean(axis=0, keepdims=True)
spec_std = X_spec_train.std(axis=0, keepdims=True)
spec_std[spec_std < 1e-6] = 1.0

temp_mean = X_temp_train.mean(axis=0, keepdims=True)
temp_std = X_temp_train.std(axis=0, keepdims=True)
temp_std[temp_std < 1e-6] = 1.0

X_spec_train_norm = (X_spec_train - spec_mean) / spec_std
X_spec_val_norm = (X_spec_val - spec_mean) / spec_std

X_temp_train_norm = (X_temp_train - temp_mean) / temp_std
X_temp_val_norm = (X_temp_val - temp_mean) / temp_std

# Save as a new artifact; do not overwrite previous experiments
out_path = experiments_dir / "aurix_specialized_features_pilot_seed13711.npz"

np.savez_compressed(
    out_path,
    X_spec_train=X_spec_train_norm.astype(np.float32),
    X_spec_val=X_spec_val_norm.astype(np.float32),
    X_temp_train=X_temp_train_norm.astype(np.float32),
    X_temp_val=X_temp_val_norm.astype(np.float32),
    train_filenames=np.array(train_names),
    val_filenames=np.array(val_names),
    spec_mean=spec_mean.astype(np.float32),
    spec_std=spec_std.astype(np.float32),
    temp_mean=temp_mean.astype(np.float32),
    temp_std=temp_std.astype(np.float32),
)

print("\n--- Results ---")
print("Spectral train:", X_spec_train_norm.shape)
print("Spectral validation:", X_spec_val_norm.shape)
print("Temporal train:", X_temp_train_norm.shape)
print("Temporal validation:", X_temp_val_norm.shape)
print("All normalized features finite:", all([
    np.isfinite(X_spec_train_norm).all(),
    np.isfinite(X_spec_val_norm).all(),
    np.isfinite(X_temp_train_norm).all(),
    np.isfinite(X_temp_val_norm).all()
]))
print("Saved:", out_path)
```

```python
from pathlib import Path
import numpy as np
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

# Reproducibility and CPU-friendly pilot
SEED = 13711
torch.manual_seed(SEED)
np.random.seed(SEED)
device = torch.device("cpu")

project_dir = Path("/content/AURIX.V1")
experiments_dir = project_dir / "experiments"
features_path = experiments_dir / "aurix_specialized_features_pilot_seed13711.npz"

assert features_path.exists(), f"Feature file not found: {features_path}"

data = np.load(features_path)

X_spec_train = data["X_spec_train"]
X_spec_val = data["X_spec_val"]
X_temp_train = data["X_temp_train"]
X_temp_val = data["X_temp_val"]

# Small autoencoder for each specialized representation
class SmallAutoencoder(nn.Module):
    def __init__(self, input_dim=128, hidden_dim=32, bottleneck_dim=8):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, bottleneck_dim),
            nn.ReLU(),
            nn.Linear(bottleneck_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, input_dim),
        )

    def forward(self, x):
        return self.net(x)

def train_one_path(X_train, X_val, path_name):
    torch.manual_seed(SEED)

    train_tensor = torch.from_numpy(
        np.asarray(X_train, dtype=np.float32)
    )
    val_tensor = torch.from_numpy(
        np.asarray(X_val, dtype=np.float32)
    )

    loader = DataLoader(
        TensorDataset(train_tensor),
        batch_size=256,
        shuffle=True,
        generator=torch.Generator().manual_seed(SEED)
    )

    model = SmallAutoencoder().to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    loss_fn = nn.MSELoss()

    # Short feasibility run, not final training
    epochs = 5

    for epoch in range(epochs):
        model.train()
        losses = []

        for (batch,) in loader:
            batch = batch.to(device)
            optimizer.zero_grad()
            reconstruction = model(batch)
            loss = loss_fn(reconstruction, batch)
            loss.backward()
            optimizer.step()
            losses.append(loss.item())

        print(
            f"{path_name} | Epoch {epoch + 1}/{epochs} "
            f"| mean train MSE: {np.mean(losses):.5f}"
        )

    model.eval()
    with torch.no_grad():
        val_reconstruction = model(val_tensor)
        val_errors = torch.mean(
            (val_reconstruction - val_tensor) ** 2,
            dim=1
        ).cpu().numpy()

    model_path = experiments_dir / f"aurix_{path_name}_pilot_seed13711.pt"
    torch.save(model.state_dict(), model_path)

    print(f"{path_name} model saved: {model_path}")
    print(f"{path_name} validation frame errors: {val_errors.shape}")
    print(f"{path_name} validation errors finite: {np.isfinite(val_errors).all()}")

    return model, val_errors

print("Training spectral pathway...")
spectral_model, spectral_val_errors = train_one_path(
    X_spec_train, X_spec_val, "spectral"
)

print("\nTraining temporal pathway...")
temporal_model, temporal_val_errors = train_one_path(
    X_temp_train, X_temp_val, "temporal"
)

print("\nPilot training finished.")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import torch

experiments_dir = Path("/content/AURIX.V1/experiments")
data = np.load(
    experiments_dir / "aurix_specialized_features_pilot_seed13711.npz"
)

def reconstruction_errors(model, X):
    model.eval()
    tensor = torch.from_numpy(np.asarray(X, dtype=np.float32))
    with torch.no_grad():
        reconstructed = model(tensor)
        errors = ((reconstructed - tensor) ** 2).mean(dim=1)
    return errors.cpu().numpy()

# Calculate frame-level errors for normal training recordings
spec_train_errors = reconstruction_errors(
    spectral_model, data["X_spec_train"]
)
temp_train_errors = reconstruction_errors(
    temporal_model, data["X_temp_train"]
)

# Recalculate validation errors consistently
spec_val_errors = reconstruction_errors(
    spectral_model, data["X_spec_val"]
)
temp_val_errors = reconstruction_errors(
    temporal_model, data["X_temp_val"]
)

# Aggregate frames into one score per recording
def recording_means(frame_errors, frames_per_recording):
    assert len(frame_errors) % frames_per_recording == 0
    return frame_errors.reshape(-1, frames_per_recording).mean(axis=1)

spec_train_recording = recording_means(spec_train_errors, 313)
temp_train_recording = recording_means(temp_train_errors, 312)
spec_val_recording = recording_means(spec_val_errors, 313)
temp_val_recording = recording_means(temp_val_errors, 312)

# Calibrate the two score scales using training scores only.
# This is exploratory calibration, not final threshold selection.
def standardize_scores(scores, reference):
    mean = reference.mean()
    std = reference.std()
    if std < 1e-8:
        std = 1.0
    return (scores - mean) / std

spec_val_z = standardize_scores(spec_val_recording, spec_train_recording)
temp_val_z = standardize_scores(temp_val_recording, temp_train_recording)

# Equal-weight fusion for inspection only
fusion_val = (spec_val_z + temp_val_z) / 2

results = pd.DataFrame({
    "filename": data["val_filenames"],
    "label": "normal",
    "spectral_score": spec_val_recording,
    "temporal_score": temp_val_recording,
    "spectral_standardized": spec_val_z,
    "temporal_standardized": temp_val_z,
    "fusion_score_exploratory": fusion_val,
})

out_path = experiments_dir / "aurix_pathway_validation_scores_seed13711.csv"
results.to_csv(out_path, index=False)

print("Validation recordings:", len(results))
print("All scores finite:", np.isfinite(
    results.select_dtypes(include=[np.number]).to_numpy()
).all())
print("\nPer-recording scores:")
display(results.round(4))
print("\nSaved:", out_path)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import librosa
import torch
from sklearn.metrics import roc_auc_score, average_precision_score

# Existing project files
exp = Path("/content/AURIX.V1/experiments")
test_dir = Path("/content/AURIX_data/bearingEmu/test")
data = np.load(exp / "aurix_specialized_features_pilot_seed13711.npz")

device = torch.device("cpu")

# Recreate the exact model architecture used during training
class SmallAutoencoder(torch.nn.Module):
    def __init__(self, input_dim=128, hidden_dim=32, bottleneck_dim=8):
        super().__init__()
        self.net = torch.nn.Sequential(
            torch.nn.Linear(input_dim, hidden_dim),
            torch.nn.ReLU(),
            torch.nn.Linear(hidden_dim, bottleneck_dim),
            torch.nn.ReLU(),
            torch.nn.Linear(bottleneck_dim, hidden_dim),
            torch.nn.ReLU(),
            torch.nn.Linear(hidden_dim, input_dim),
        )

    def forward(self, x):
        return self.net(x)

spectral_model = SmallAutoencoder()
spectral_model.load_state_dict(
    torch.load(exp / "aurix_spectral_pilot_seed13711.pt",
               map_location="cpu", weights_only=True)
)
spectral_model.eval()

temporal_model = SmallAutoencoder()
temporal_model.load_state_dict(
    torch.load(exp / "aurix_temporal_pilot_seed13711.pt",
               map_location="cpu", weights_only=True)
)
temporal_model.eval()

# Extract features and apply training-only normalization
def features_for_file(path):
    audio, sr = librosa.load(str(path), sr=16000, mono=False)
    ch0 = audio[0] if audio.ndim == 2 else audio

    mel = librosa.feature.melspectrogram(
        y=ch0, sr=sr, n_fft=1024, hop_length=512,
        n_mels=128, power=2.0, fmin=0, fmax=8000
    )
    log_mel = 10 * np.log10(
        np.maximum(mel, np.finfo(np.float32).eps)
    ).astype(np.float32)

    spectral = log_mel.T
    temporal = np.diff(log_mel, axis=1).T

    spectral = (spectral - data["spec_mean"]) / data["spec_std"]
    temporal = (temporal - data["temp_mean"]) / data["temp_std"]

    return spectral.astype(np.float32), temporal.astype(np.float32)

def frame_errors(model, X):
    with torch.no_grad():
        x = torch.from_numpy(X)
        reconstructed = model(x)
        return ((reconstructed - x) ** 2).mean(dim=1).numpy()

# Calibrate score scales using only the pilot's normal training recordings
spec_train_errors = frame_errors(spectral_model, data["X_spec_train"])
temp_train_errors = frame_errors(temporal_model, data["X_temp_train"])

spec_train_scores = spec_train_errors.reshape(-1, 313).mean(axis=1)
temp_train_scores = temp_train_errors.reshape(-1, 312).mean(axis=1)

def zscore(scores, reference):
    std = reference.std()
    if std < 1e-8:
        std = 1.0
    return (scores - reference.mean()) / std

# Score every test recording
rows = []
files = sorted(test_dir.glob("*.wav"))
assert files, f"No WAV files found in {test_dir}"

for i, path in enumerate(files, start=1):
    spec, temp = features_for_file(path)

    spec_score = frame_errors(spectral_model, spec).mean()
    temp_score = frame_errors(temporal_model, temp).mean()

    spec_z = zscore(np.array([spec_score]), spec_train_scores)[0]
    temp_z = zscore(np.array([temp_score]), temp_train_scores)[0]

    # Labels are parsed from the dataset filenames
    label = 1 if "_anomaly_" in path.name else 0

    rows.append({
        "filename": path.name,
        "label": label,
        "spectral_score": spec_score,
        "temporal_score": temp_score,
        "spectral_z": spec_z,
        "temporal_z": temp_z,
        "fusion_score": (spec_z + temp_z) / 2,
    })

    if i % 25 == 0 or i == len(files):
        print(f"Scored {i}/{len(files)} recordings")

results = pd.DataFrame(rows)
assert results["label"].nunique() == 2, "Expected normal and anomaly labels"

# Save without overwriting previous experiments
scores_path = exp / "aurix_specialized_test_scores_exploratory_seed13711.csv"
results.to_csv(scores_path, index=False)

print("\nTest label counts:")
print(results["label"].map({0: "normal", 1: "anomaly"}).value_counts())

print("\nExploratory metrics (higher is better):")
for name, column in [
    ("Spectral", "spectral_score"),
    ("Temporal", "temporal_score"),
    ("Fusion", "fusion_score"),
]:
    auc = roc_auc_score(results["label"], results[column])
    ap = average_precision_score(results["label"], results[column])
    print(f"{name:10s} ROC-AUC={auc:.4f} | Average Precision={ap:.4f}")

print("\nSaved scores:", scores_path)
print("All numeric scores finite:",
      np.isfinite(results.select_dtypes(include=[np.number]).to_numpy()).all())
```

```python
import numpy as np
import torch
import torch.nn as nn
from pathlib import Path

root = Path("/content/AURIX.V1/experiments")

data = np.load(root / "aurix_specialized_features_pilot_seed13711.npz")

class SmallAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(128, 32),
            nn.ReLU(),
            nn.Linear(32, 8),
            nn.ReLU(),
            nn.Linear(8, 32),
            nn.ReLU(),
            nn.Linear(32, 128),
        )

    def forward(self, x):
        return self.net(x)

def recording_errors(features, model, frames_per_recording):
    model.eval()
    with torch.no_grad():
        x = torch.as_tensor(features, dtype=torch.float32)
        frame_errors = ((model(x) - x) ** 2).mean(dim=1).numpy()

    n = len(frame_errors) // frames_per_recording
    if n * frames_per_recording != len(frame_errors):
        raise ValueError("عدد الـ frames مش متوافق مع طول التسجيلات.")

    return frame_errors.reshape(n, frames_per_recording).mean(axis=1)

models = {
    "Spectral": (
        "aurix_spectral_pilot_seed13711.pt",
        "X_spec_train",
        313,
    ),
    "Temporal": (
        "aurix_temporal_pilot_seed13711.pt",
        "X_temp_train",
        312,
    ),
}

for name, (filename, feature_key, frames) in models.items():
    model = SmallAutoencoder()
    model.load_state_dict(
        torch.load(root / filename, map_location="cpu", weights_only=True)
    )

    scores = recording_errors(data[feature_key], model, frames)

    print(f"\n--- {name} training scores ---")
    print("Number of recordings:", len(scores))
    print("Mean:", round(float(scores.mean()), 6))
    print("Std (ddof=0):", round(float(scores.std()), 6))
    print("Std (ddof=1):", round(float(scores.std(ddof=1)), 6))
    print("Min:", round(float(scores.min()), 6))
    print("Median:", round(float(np.median(scores)), 6))
    print("Max:", round(float(scores.max()), 6))
    print("All finite:", bool(np.isfinite(scores).all()))
```
